```python
import pandas as pd

demo = pd.read_csv('../data/demographic.csv', usecols = ['SEQN', 'RIDRETH3', 'RIDAGEYR'])
exam = pd.read_csv('../data/examination.csv', usecols = ['SEQN', 'BMXBMI'])
labs = pd.read_csv('../data/labs.csv', usecols = ['SEQN', 'LBXGH'])
quest = pd.read_csv('../data/questionnaire.csv', usecols = ['SEQN', 'DIQ010'])
```

```python
df = demo.merge(exam, on = 'SEQN', how = 'left')
df = df.merge(labs, on = 'SEQN', how = 'left')
df = df.merge(quest, on = 'SEQN', how = 'left')
```

```python
print(df.shape)
```

```python
df.isnull().sum() 
```

```python
#cleaning

df_analysis = df.dropna(subset = ['RIDRETH3', 'BMXBMI', 'LBXGH']) 
# Removes rows that are missing values for any of the 3 variables necessary for analysis
print(df_analysis.shape)
```

```python
#naming codes from RIDRETH3 for readability

race_labels = {
    1: 'Mexican American',
    2: 'Other Hispanic',
    3: 'Non-Hispanic White',
    4: 'Non-Hispanic Black',
    6: 'Non-Hispanic Asian',
    7: 'Other/Multiracial'
}

df_analysis['race_ethnicity'] = df_analysis['RIDRETH3'].map(race_labels)

print(df_analysis)
```

```python
#using hba1c to label diabetes status

def classify_diabetes(hba1c):
    if pd.isna(hba1c):
        return None
    elif hba1c < 5.7:
        return 'Normal'
    elif hba1c < 6.5:
        return 'Prediabetes'
    else:
        return 'Diabetes'

df_analysis['diabetes_status'] = df_analysis['LBXGH'].apply(classify_diabetes)

print(df_analysis)
```

```python
df_analysis.to_csv('../data/cleaned_merged.csv', index = False)
```
